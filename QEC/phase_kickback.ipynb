```python
from qiskit.circuit.library import HGate, XGate, CXGate
from qiskit.circuit import QuantumCircuit, QuantumRegister
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

"""The below circuit is implemented to showcase teh concept of pashekickback by using the deutch algorithm.
In this circuit, we will see how the phase of the qubit is changed based on the function implemented by the oracle.
"""

# Defining a register
quantum_reg = QuantumRegister(2, 'q')

# Creating a Qunatum Circuit
circuit = QuantumCircuit(quantum_reg)
circuit.append(HGate(), [0])
circuit.append(XGate(), [1])
state_a = Statevector.from_instruction(circuit)
circuit.append(HGate(), [1]) 
state_b = Statevector.from_instruction(circuit)
circuit.append(CXGate(), [0,1])
state_c = Statevector.from_instruction(circuit)
circuit.draw("mpl")
```
